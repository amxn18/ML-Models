```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
%matplotlib inline
```

```python
trainDF = pd.read_excel('dataset/Data_Train.xlsx')
trainDF.head()
```

```python
testDF = pd.read_excel('dataset/Test_set.xlsx')
testDF.head()
```

```python
df = pd.concat([trainDF, testDF], axis = 0)
print(df.shape)
df.head()
```

```python
df.info()
```

```python
df['Date'] = df['Date_of_Journey'].str.split('/').str[0].astype('int64')
df['Month'] = df['Date_of_Journey'].str.split('/').str[1].astype('int64')
df['Year'] = df['Date_of_Journey'].str.split('/').str[2].astype('int64')
df = df.drop(columns=['Date_of_Journey'])
df.head()
```

```python
df.info()
```

```python
df['Arrival_Time'] = df['Arrival_Time'].str.split(' ').str[0]
df.head()
```

```python
df['Arrival_Hours'] = df['Arrival_Time'].str.split(':').str[0].astype('int64')
df['Arrival_Minutes'] = df['Arrival_Time'].str.split(':').str[1].astype('int64')
df = df.drop(columns= ['Arrival_Time'])
df.head()
```

```python
df['Departure_Hours'] = df['Dep_Time'].str.split(':').str[0].astype('int64')
df['Departure_Minutes'] = df['Dep_Time'].str.split(':').str[1].astype('int64')
df = df.drop(columns= ['Dep_Time'])
```

```python
df.head()
```

```python
df.info()
```

```python
df['Additional_Info'].unique()
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset=['Route'])
df.isnull().sum()
```

```python
df['Total_Stops'].unique()
```

```python
df['Total_Stops'] = df['Total_Stops'].map({'non-stop': 0, '1 stop': 1, '2 stops': 2, '4 stops': 3})
```

```python
df = df.drop(columns=['Route'])
```

```python
def convert_duration(x):
    x = x.split()
    
    hours = 0
    minutes = 0
    
    if len(x) == 2:
        hours = int(x[0][:-1])
        minutes = int(x[1][:-1])
    elif 'h' in x[0]:
        hours = int(x[0][:-1])
    else:
        minutes = int(x[0][:-1])
    
    return hours * 60 + minutes

df['Duration'] = df['Duration'].apply(convert_duration)
```

```python
df.head()
```

```python
df['Airline'].unique()
```

```python
from sklearn.preprocessing import LabelEncoder
Encoder = LabelEncoder()
df['Airline'] = Encoder.fit_transform(df['Airline'])
df['Source'] = Encoder.fit_transform(df['Source'])
df['Destination'] = Encoder.fit_transform(df['Destination'])
df['Additional_Info'] = Encoder.fit_transform(df['Additional_Info'])
```
